```python
# -*- coding: utf-8 -*-
"""
confound_analysis_by_firing.py

Correct dead-feature check for TopK SAEs with a continuously unit-normalized
decoder: decoder norm cannot distinguish dead from live features here, since
normalize_decoder() rescales every column to unit length after every training
step regardless of whether that feature ever fired. A feature is dead if and
only if it never enters the top-k set on real data -- this script measures
that directly, by running each checkpoint's SAE over a batch of held-out
Pythia-70m activations and counting, per feature, how many tokens it fires on.

Requires transformers and datasets (no transformer_lens dependency -- uses plain
HF output_hidden_states instead, for compatibility with older Python versions).
"""

import re
import numpy as np
import torch
import torch.nn as nn
from pathlib import Path
from typing import Dict, List, Tuple
from transformers import AutoModel, AutoTokenizer
from datasets import load_dataset

CHECKPOINT_DIR = "checkpoints"
SEEDS = [42, 256, 1024]
KS = [64, 128, 256]
TOKEN_COUNTS = [250_000_000, 500_000_000, 1_000_000_000]  # matches the checkpoints checked earlier

N_EVAL_BATCHES = 40       # reduced from 40 to cut memory footprint; raise once stable
NEVER_FIRED_THRESHOLD = 0  # a feature with 0 firings in the eval set is dead
RARE_FIRING_THRESHOLD = 10  # features firing fewer than this many times are flagged as
                             # "barely alive" -- not dead, but too undertrained to trust


class SparseAutoencoder(nn.Module):
    """Reconstructs the forward pass from the checkpoint's saved weights. TopK is
    applied directly to the pre-activation (no ReLU), matching the confirmed
    architecture used to produce these checkpoints."""

    def __init__(self, d_model: int, n_features: int, k: int):
        super().__init__()
        self.d_model = d_model
        self.n_features = n_features
        self.k = k
        self.W_enc = nn.Parameter(torch.zeros(d_model, n_features))
        self.b_enc = nn.Parameter(torch.zeros(n_features))
        self.W_dec = nn.Parameter(torch.zeros(n_features, d_model))
        self.b_dec = nn.Parameter(torch.zeros(d_model))

    def encode(self, x: torch.Tensor) -> torch.Tensor:
        pre_acts = (x - self.b_dec) @ self.W_enc + self.b_enc
        topk_vals, topk_idx = pre_acts.topk(self.k, dim=-1)
        f = torch.zeros_like(pre_acts)
        f.scatter_(-1, topk_idx, topk_vals)
        return f


def load_sae(path: Path) -> Tuple[SparseAutoencoder, dict]:
    ckpt = torch.load(path, map_location="cpu")
    cfg = ckpt["config"]
    k = ckpt.get("k", cfg.get("k"))
    if k is None:
        raise ValueError(f"Could not find k in checkpoint or config for {path}")
    sae = SparseAutoencoder(cfg["d_model"], cfg["n_features"], k=k)
    sae.load_state_dict(ckpt["model_state_dict"])
    return sae, cfg


def load_pythia(model_name: str, device: str):
    """Load Pythia via plain transformers -- no transformer_lens needed. Uses
    AutoModel (the base transformer, no language-modeling head) rather than
    AutoModelForCausalLM: the lm_head projects every token to the ~50k-token
    vocabulary, which we never use here and which was the actual source of the
    OOM (256 * 128 * 50304 * 4 bytes = 6.6GB per batch, just for logits nobody
    reads). Returns (model, tokenizer)."""
    hf_name = model_name if "/" in model_name else f"EleutherAI/{model_name}"
    tokenizer = AutoTokenizer.from_pretrained(hf_name)
    try:
        model = AutoModel.from_pretrained(hf_name, dtype=torch.float32)
    except TypeError:  # older transformers spells this torch_dtype
        model = AutoModel.from_pretrained(hf_name, torch_dtype=torch.float32)
    model.to(device)
    model.eval()
    return model, tokenizer


def build_eval_activations(model, tokenizer, hook_point: str, seq_len: int, batch_size: int,
                            device: str, n_batches: int) -> torch.Tensor:
    """Same streaming setup as the training script's eval set: fresh read from
    the start of the same dataset, so results are directly comparable to what
    the original pipeline measured.

    hook_point looks like 'blocks.3.hook_resid_post' in transformer_lens naming.
    In plain HF's output_hidden_states, hidden_states[0] is the embedding output
    and hidden_states[i] is the residual stream AFTER layer i (i=1..n_layers) --
    exactly what 'blocks.{i-1}.hook_resid_post' means in transformer_lens terms.
    So layer L in the hook_point string corresponds to hidden_states[L + 1].
    """
    layer_idx = int(hook_point.split(".")[1])
    hs_index = layer_idx + 1

    dataset = load_dataset("monology/pile-uncopyrighted", split="train", streaming=True)
    token_buffer = []
    batches = []

    for example in dataset:
        tokens = tokenizer(
            example["text"], return_tensors="pt", truncation=True, max_length=seq_len * 10
        )["input_ids"][0]
        token_buffer.extend(tokens.tolist())

        while len(token_buffer) >= seq_len * batch_size:
            batch_tokens = torch.tensor(
                token_buffer[: seq_len * batch_size]
            ).reshape(batch_size, seq_len)
            token_buffer = token_buffer[seq_len * batch_size:]

            with torch.no_grad():
                out = model(batch_tokens.to(device), output_hidden_states=True)
                acts = out.hidden_states[hs_index]  # (batch_size, seq_len, d_model)
                acts = acts.reshape(-1, acts.shape[-1]).cpu()
            batches.append(acts)

            if len(batches) >= n_batches:
                return torch.cat(batches, dim=0)

    return torch.cat(batches, dim=0)


def compute_firing_counts(sae: SparseAutoencoder, activations: torch.Tensor,
                           device: str, batch_size: int = 4096) -> np.ndarray:
    """Count, per feature, how many tokens it enters the top-k set on."""
    sae = sae.to(device)
    sae.eval()
    counts = torch.zeros(sae.n_features, device=device)

    with torch.no_grad():
        for start in range(0, len(activations), batch_size):
            batch = activations[start:start + batch_size].to(device)
            f = sae.encode(batch)
            counts += (f != 0).sum(dim=0).float()

    return counts.cpu().numpy()


def main():
    ckpt_dir = Path(CHECKPOINT_DIR)

    # All checkpoints in this sweep share the same model/hook point/architecture,
    # so we only need to build the eval activations once.
    sample_path = None
    for tokens in TOKEN_COUNTS:
        for k in KS:
            p = ckpt_dir / f"seed{SEEDS[0]}_k{k}_tokens{tokens}.pt"
            if p.exists():
                sample_path = p
                break
        if sample_path:
            break
    if sample_path is None:
        print("No checkpoints found matching the expected naming pattern -- check CHECKPOINT_DIR.")
        return

    sample_ckpt = torch.load(sample_path, map_location="cpu")
    cfg = sample_ckpt["config"]
    model_name = cfg.get("model_name", "pythia-70m-deduped")
    hook_point = cfg["hook_point"]
    seq_len = cfg.get("seq_len", 128)
    batch_size = cfg.get("batch_size", 256)
    device = "cuda" if torch.cuda.is_available() else "cpu"

    print(f"Loading {model_name} to generate eval activations at {hook_point}...")
    model, tokenizer = load_pythia(model_name, device)

    print(f"Building eval set ({N_EVAL_BATCHES} batches, "
          f"{N_EVAL_BATCHES * batch_size * seq_len:,} tokens)...")
    activations = build_eval_activations(
        model, tokenizer, hook_point, seq_len, batch_size, device, N_EVAL_BATCHES
    )
    print(f"Eval activations: {activations.shape}\n")

    print("=" * 100)
    print("Dead-feature check via true firing counts (top-k entry on held-out data)")
    print("=" * 100)
    header = (f"{'k':>5} {'tokens':>13} {'seed':>6} {'n_features':>10} "
              f"{'%never-fired':>13} {'%rarely-fired':>14} {'median firings':>15}")
    print(header)

    results = []
    for tokens in TOKEN_COUNTS:
        for k in KS:
            for seed in SEEDS:
                path = ckpt_dir / f"seed{seed}_k{k}_tokens{tokens}.pt"
                if not path.exists():
                    continue
                sae, cfg = load_sae(path)
                counts = compute_firing_counts(sae, activations, device)

                n_features = sae.n_features
                never_fired = int((counts <= NEVER_FIRED_THRESHOLD).sum())
                rarely_fired = int((counts < RARE_FIRING_THRESHOLD).sum())
                median_firings = float(np.median(counts))

                results.append({
                    "k": k, "tokens": tokens, "seed": seed,
                    "n_features": n_features,
                    "pct_never_fired": 100 * never_fired / n_features,
                    "pct_rarely_fired": 100 * rarely_fired / n_features,
                    "median_firings": median_firings,
                    "counts": counts,
                })
                print(f"{k:>5} {tokens:>13,} {seed:>6} {n_features:>10} "
                      f"{100 * never_fired / n_features:>12.1f}% "
                      f"{100 * rarely_fired / n_features:>13.1f}% "
                      f"{median_firings:>15,.0f}")

    if not results:
        print("\nNo matching checkpoints found for the configured TOKEN_COUNTS/KS/SEEDS.")
        return

    print()
    print("Interpretation:")
    print("  - %never-fired is the real dead-feature rate. Compare this against the earlier")
    print("    decoder-norm-based check, which reported 0.0% everywhere -- if %never-fired is")
    print("    substantially higher, that confirms the norm-based check was invalid here and")
    print("    the true dead-feature rate needs to replace it in the confound-checks section.")
    print("  - %rarely-fired (< 10 firings in this eval set) flags undertrained features that")
    print("    aren't technically dead but may still behave unpredictably in matching.")
    print("  - If dead/rare rates rise sharply with k, that revives the undertraining")
    print("    explanation for the k=256 result and would need to be reconciled with, or may")
    print("    replace, the capacity-based account in the discussion section.")

    # Save per-feature counts for further analysis (e.g. cross-referencing dead features
    # against which ones were labeled "unstable" by the reappearance-probability metric)
    Path("outputs").mkdir(exist_ok=True)
    np.savez(
        "outputs/firing_counts.npz",
        **{f"k{r['k']}_seed{r['seed']}_tok{r['tokens']}": r["counts"] for r in results}
    )
    print("\nSaved per-feature firing counts to outputs/firing_counts.npz")


if __name__ == "__main__":
    main()
```

```python
# -*- coding: utf-8 -*-
"""
stability_live_only.py

Recomputes cross-seed feature stability restricted to LIVE features only
(features that fired at least once in the held-out eval set), using the
per-feature firing_counts.npz produced by confound_analysis.py.

Motivation: k=256 was found to have a severe, growing dead-feature problem
(57.7% -> 70.4% -> 72-73% never-fired across checkpoints, with median firing
count of 0). Since a dead feature's decoder direction is frozen wherever it
stopped receiving gradient, it cannot possibly match across seeds -- this
would trivially depress the stability metric regardless of any geometric
effect. This script answers the decisive question: does restricting the
stability computation to only live features close most of the gap between
k=256 and k=64/128, or does a real gap remain even among live features?

Reappearance-probability computation (many-to-one argmax matching, theta=0.7)
is unchanged from the original pipeline -- only the feature POPULATION being
scored changes (all features vs. live-only features).
"""

import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
from pathlib import Path
from typing import Dict, Tuple

CHECKPOINT_DIR = "checkpoints"  # matches the Colab download location
FIRING_COUNTS_PATH = "outputs/firing_counts.npz"
SEEDS = [42, 256, 1024]
KS = [64, 128, 256]
TOKEN_COUNTS = [250_000_000, 500_000_000, 1_000_000_000]

THETA = 0.7    # matches the original stability metric's matching threshold
EPSILON = 0.05  # matches the original endpoint-binarization epsilon
MIN_FIRINGS = 1  # a feature with 0 firings in the eval set is "dead"; this is
                 # the floor for counting as "live" -- raise to e.g. 10 to also
                 # exclude barely-alive features if you want a stricter cut


class SparseAutoencoder(nn.Module):
    """Same reconstruction as confound_analysis.py -- only W_dec is used here."""

    def __init__(self, d_model: int, n_features: int, k: int):
        super().__init__()
        self.d_model = d_model
        self.n_features = n_features
        self.k = k
        self.W_enc = nn.Parameter(torch.zeros(d_model, n_features))
        self.b_enc = nn.Parameter(torch.zeros(n_features))
        self.W_dec = nn.Parameter(torch.zeros(n_features, d_model))
        self.b_dec = nn.Parameter(torch.zeros(d_model))


def load_sae(path: Path) -> SparseAutoencoder:
    ckpt = torch.load(path, map_location="cpu", weights_only=False)
    cfg = ckpt["config"]
    k = ckpt.get("k", cfg.get("k"))
    sae = SparseAutoencoder(cfg["d_model"], cfg["n_features"], k=k)
    sae.load_state_dict(ckpt["model_state_dict"])
    return sae


def compute_decoder_similarity(sae1: SparseAutoencoder, sae2: SparseAutoencoder) -> torch.Tensor:
    W1 = F.normalize(sae1.W_dec.detach(), dim=1)
    W2 = F.normalize(sae2.W_dec.detach(), dim=1)
    return (W1 @ W2.T).cpu()


def compute_reappearance_probability(
    saes: Dict[int, SparseAutoencoder], theta: float
) -> Tuple[np.ndarray, np.ndarray]:
    """Many-to-one argmax matching, unchanged from the original pipeline.
    Returns (reappearance_probs, best_sim_first_comparison) for the anchor seed."""
    seeds = list(saes.keys())
    anchor_sae = saes[seeds[0]]
    n_features = anchor_sae.n_features
    reappearance_counts = np.zeros(n_features)
    first_best_sim = None

    for i, other_seed in enumerate(seeds[1:]):
        sim_matrix = compute_decoder_similarity(anchor_sae, saes[other_seed])
        best_sim, _ = sim_matrix.max(dim=1)
        best_sim = best_sim.numpy()
        if i == 0:
            first_best_sim = best_sim
        reappearance_counts += (best_sim >= theta).astype(float)

    N = len(seeds) - 1
    return reappearance_counts / N, first_best_sim


def main():
    ckpt_dir = Path(CHECKPOINT_DIR)
    firing_data = np.load(FIRING_COUNTS_PATH)

    print("=" * 100)
    print("Stability comparison: ALL features vs. LIVE-ONLY features")
    print("=" * 100)
    header = (f"{'k':>5} {'tokens':>13} {'n_live':>8} {'%live':>7} "
              f"{'%stable(all)':>13} {'%stable(live)':>14}")
    print(header)

    results = []
    for tokens in TOKEN_COUNTS:
        for k in KS:
            paths = {s: ckpt_dir / f"seed{s}_k{k}_tokens{tokens}.pt" for s in SEEDS}
            if not all(p.exists() for p in paths.values()):
                print(f"k={k}, tokens={tokens:,}: missing checkpoints, skipping.")
                continue

            saes = {s: load_sae(p) for s, p in paths.items()}
            anchor_seed = SEEDS[0]

            # Reappearance probability over ALL features (the original metric)
            reappearance_probs, _ = compute_reappearance_probability(saes, THETA)
            stable_all = (reappearance_probs >= (1 - EPSILON))
            pct_stable_all = 100 * stable_all.mean()

            # Live mask: use the anchor seed's firing counts (matches MIN_FIRINGS
            # convention -- we're asking "of the features THIS SAE actually uses,
            # how many are stable", not filtering on the comparison SAEs' liveness)
            firing_key = f"k{k}_seed{anchor_seed}_tok{tokens}"
            if firing_key not in firing_data:
                print(f"k={k}, tokens={tokens:,}: no firing counts for anchor seed "
                      f"{anchor_seed} (key {firing_key!r} not in {FIRING_COUNTS_PATH}), skipping.")
                continue
            firing_counts = firing_data[firing_key]
            live_mask = firing_counts >= MIN_FIRINGS

            n_live = int(live_mask.sum())
            pct_live = 100 * live_mask.mean()

            if n_live == 0:
                pct_stable_live = float("nan")
            else:
                pct_stable_live = 100 * stable_all[live_mask].mean()

            results.append({
                "k": k, "tokens": tokens,
                "n_live": n_live, "pct_live": pct_live,
                "pct_stable_all": pct_stable_all, "pct_stable_live": pct_stable_live,
            })
            print(f"{k:>5} {tokens:>13,} {n_live:>8} {pct_live:>6.1f}% "
                  f"{pct_stable_all:>12.1f}% {pct_stable_live:>13.1f}%")

    if not results:
        print("\nNo results computed -- check CHECKPOINT_DIR and FIRING_COUNTS_PATH.")
        return

    print()
    print("Interpretation:")
    print("  - %stable(all) is the original headline number (dead features included,")
    print("    counted as automatically unstable).")
    print("  - %stable(live) restricts to only features that fired at least once in the")
    print("    held-out eval set -- this is the fair comparison, since a dead feature")
    print("    cannot possibly match across seeds regardless of any real effect.")
    print("  - If %stable(live) for k=256 rises to something close to k=64/128's numbers,")
    print("    dead features explain most or all of the k=256 result -- the capacity/")
    print("    dimensionality hypothesis is likely unnecessary.")
    print("  - If a large gap remains between k=256 and k=64/128 even restricted to live")
    print("    features, there is a real residual instability effect beyond dead features,")
    print("    and the capacity hypothesis (or another mechanism) is still worth discussing.")

    # Save for reference / plotting
    import json
    Path("outputs").mkdir(exist_ok=True)
    with open("outputs/stability_live_only.json", "w") as f:
        json.dump(results, f, indent=2, default=float)
    print("\nSaved outputs/stability_live_only.json")


if __name__ == "__main__":
    main()
```

```python
# -*- coding: utf-8 -*-
"""
check_near_duplicates_live_only.py

Recomputes the within-SAE near-duplicate (redundancy) check restricted to
LIVE features only, using the firing_counts.npz produced by
confound_analysis.py.

Motivation: the original near-duplicate check (diagnose_k256.py) compared
cosine similarity between ALL decoder columns, including the ~70%+ of
features that are dead at k=256. Dead features are frozen at their (tied)
initialization -- in a 512-dim space this is very likely an effectively
random direction relative to everything else, which would almost never
register as a near-duplicate of another feature. If so, the large dead
population could be DILUTING the apparent redundancy rate, masking real
redundancy that might exist specifically among the live features.

This isn't fixing an invalid check the way the dead-feature check needed
fixing -- cosine similarity between decoder columns is unaffected by the
unit-norm constraint, since it compares relative angles, not absolute norms.
This is a population-composition check: does redundancy look different when
computed only over the features that actually matter (i.e. fire)?
"""

import numpy as np
import torch
import torch.nn.functional as F
from pathlib import Path

CHECKPOINT_DIR = "checkpoints"  # matches the Colab download location
FIRING_COUNTS_PATH = "outputs/firing_counts.npz"
SEEDS = [42, 256, 1024]
KS = [64, 128, 256]
TOKEN_COUNTS = [250_000_000, 500_000_000, 1_000_000_000]

DUPLICATE_THETA = 0.9  # matches the original diagnose_k256.py threshold
MIN_FIRINGS = 1         # same live-feature definition as stability_live_only.py


def load_decoder_weights(path: Path) -> torch.Tensor:
    ckpt = torch.load(path, map_location="cpu", weights_only=False)
    return ckpt["model_state_dict"]["W_dec"]  # [n_features, d_model]


def near_duplicate_stats(W_dec: torch.Tensor, live_mask: np.ndarray, theta: float) -> dict:
    """Compute near-duplicate rate over all features vs. live features only.
    'Near-duplicate' means the closest OTHER feature (by cosine similarity)
    exceeds theta. When restricting to live-only, candidates are also
    restricted to live features -- a live feature's nearest neighbor should
    be assessed against other live features, not against frozen dead ones."""
    W_norm = F.normalize(W_dec, dim=1)
    sim = W_norm @ W_norm.T
    sim.fill_diagonal_(-1.0)

    # All-features version (matches the original diagnose_k256.py check)
    max_sim_all, _ = sim.max(dim=1)
    near_dup_all = (max_sim_all >= theta).numpy()

    # Live-only version: restrict both the features being scored AND their
    # candidate neighbors to the live set.
    live_idx = np.where(live_mask)[0]
    if len(live_idx) < 2:
        return {
            "n_live": len(live_idx),
            "pct_near_dup_all": 100 * near_dup_all.mean(),
            "pct_near_dup_live": float("nan"),
            "mean_max_sim_all": max_sim_all.mean().item(),
            "mean_max_sim_live": float("nan"),
        }

    live_idx_t = torch.from_numpy(live_idx)
    sim_live = sim[live_idx_t][:, live_idx_t]  # [n_live, n_live] submatrix
    max_sim_live, _ = sim_live.max(dim=1)
    near_dup_live = (max_sim_live >= theta).numpy()

    return {
        "n_live": len(live_idx),
        "pct_near_dup_all": 100 * near_dup_all.mean(),
        "pct_near_dup_live": 100 * near_dup_live.mean(),
        "mean_max_sim_all": max_sim_all.mean().item(),
        "mean_max_sim_live": max_sim_live.mean().item(),
    }


def main():
    ckpt_dir = Path(CHECKPOINT_DIR)
    firing_data = np.load(FIRING_COUNTS_PATH)

    print("=" * 100)
    print("Near-duplicate (redundancy) check: ALL features vs. LIVE-ONLY features")
    print("=" * 100)
    header = (f"{'k':>5} {'tokens':>13} {'seed':>6} {'n_live':>8} "
              f"{'%near-dup(all)':>15} {'%near-dup(live)':>16} "
              f"{'mean-sim(all)':>14} {'mean-sim(live)':>15}")
    print(header)

    results = []
    for tokens in TOKEN_COUNTS:
        for k in KS:
            for seed in SEEDS:
                path = ckpt_dir / f"seed{seed}_k{k}_tokens{tokens}.pt"
                if not path.exists():
                    continue

                firing_key = f"k{k}_seed{seed}_tok{tokens}"
                if firing_key not in firing_data:
                    print(f"k={k}, tokens={tokens:,}, seed={seed}: no firing counts "
                          f"(key {firing_key!r} not found), skipping.")
                    continue

                W_dec = load_decoder_weights(path)
                firing_counts = firing_data[firing_key]
                live_mask = firing_counts >= MIN_FIRINGS

                stats = near_duplicate_stats(W_dec, live_mask, DUPLICATE_THETA)
                stats.update({"k": k, "tokens": tokens, "seed": seed})
                results.append(stats)

                print(f"{k:>5} {tokens:>13,} {seed:>6} {stats['n_live']:>8} "
                      f"{stats['pct_near_dup_all']:>14.2f}% "
                      f"{stats['pct_near_dup_live']:>15.2f}% "
                      f"{stats['mean_max_sim_all']:>14.4f} "
                      f"{stats['mean_max_sim_live']:>15.4f}")

    if not results:
        print("\nNo results computed -- check CHECKPOINT_DIR and FIRING_COUNTS_PATH.")
        return

    print()
    print("Interpretation:")
    print("  - %near-dup(all) reproduces the original diagnose_k256.py check (~0% everywhere).")
    print("  - %near-dup(live) restricts both the scored features and their candidate")
    print("    neighbors to only live features -- this tests whether dead features were")
    print("    diluting the apparent redundancy rate by acting as random, non-matching")
    print("    'padding' in the similarity search.")
    print("  - If %near-dup(live) is still ~0% at k=256, redundancy is confirmed ruled out")
    print("    even among the features that matter -- the live-feature instability found")
    print("    in stability_live_only.py is NOT explained by internal duplication either.")
    print("  - If %near-dup(live) is notably higher than %near-dup(all), that would revive")
    print("    redundancy as a partial explanation for the live-feature instability, and")
    print("    would need to be reconciled with the capacity-based account.")

    import json
    Path("outputs").mkdir(exist_ok=True)
    with open("outputs/near_duplicates_live_only.json", "w") as f:
        json.dump(results, f, indent=2, default=float)
    print("\nSaved outputs/near_duplicates_live_only.json")


if __name__ == "__main__":
    main()
```
