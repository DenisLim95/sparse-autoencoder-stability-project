```python
import os
from itertools import product

output_dir = "."

seeds = [42, 256, 1024]
ks = [64, 128, 256]
tokens = [50_000_000, 100_000_000, 250_000_000, 500_000_000, 800_000_000, 1_000_000_000]

existing = set(os.listdir(output_dir))

expected = []
missing = []

for seed, k, tok in product(seeds, ks, tokens):
    fname = f"seed{seed}_k{k}_tokens{tok}.pt"
    expected.append(fname)
    if fname not in existing:
        missing.append(fname)

print(f"Expected: {len(expected)}")
print(f"Found:    {len(expected) - len(missing)}")
print(f"Missing:  {len(missing)}")

if missing:
    print("\nMissing files:")
    for f in missing:
        print(" ", f)

# bonus: catch unexpected files too (typos, extra runs, etc.)
unexpected = existing - set(expected) - {"Unconfirmed 202882.crdownload"}
if unexpected:
    print("\nUnexpected files in directory (not matching naming pattern):")
    for f in unexpected:
        print(" ", f)
```
